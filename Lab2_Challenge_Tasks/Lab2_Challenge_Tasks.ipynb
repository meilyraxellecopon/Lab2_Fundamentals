```python
# Exercise 1: Redaction Protocol
# Surname should be uppercase and vowels replaced by the seed number.

surname = "COPON"
SEED_NUM = int(STUDENT_ID[-1])
redacted_name = ""

for ch in surname:
    if ch in "AEIOU":
        redacted_name += str(SEED_NUM)
    else:
        redacted_name += ch

print(f"Original Surname: {surname}")
print(f"Redacted Name: {redacted_name}")
```

```python
# Exercise 2: Verification Hash
# Calculate using ASCII value, seed number, and total loop cycles.

ENGINEER = "Copon"
first_letter = ENGINEER[0]
ASCII_value = ord(first_letter)
loops = SEED_NUM + 3
verification_hash = (ASCII_value * SEED_NUM) - loops

print(f"Engineer: {ENGINEER}")
print(f"First Letter: {first_letter}")
print(f"ASCII Value: {ASCII_value}")
print(f"Seed Number: {SEED_NUM}")
print(f"Loop Cycles: {loops}")
print(f"Verification Hash: {verification_hash}")
```
